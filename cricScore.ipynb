```python
class cricketscore:
    #constructor to initialize the class
    def __init__(self,rootWindow):# self trying to represnt the class object or instance of the class
        self.rootWindow = rootWindow
        self.rootWindow.title("Live Cricket Score")
        self.rootWindow.geometry('800x500')
        self.bg = ImageTk.PhotoImage(file = "cricket.jpg") #SETTING THE BACKGROUND IMAGE
        self.bg_label = Label(self.rootWindow, image = self.bg)
        self.bg_label.place(x = 0, y = 0)

        #Let's give a label to my GUI
        self.label = Label(self.rootWindow, text = "Live Cricket Score", font = ("times new roman", 60), compound = 'center')
        self.label.pack(padx=100, pady = 50)
        
        
        #fetch the live match detials and set it in my GUI
        self.var = StringVar()
        #matches kaa saara data dump krna hai
        self.matches = self.match_details() #self.match_details is a function which will return the match details
        #self.matches is a dictionary which will contain the match details



        #button to check the details of the match

    #we will create a function to fetch the live match details from cricbuzz
    def match_details(self): #method ke andar self ko pass karna padta hai kyuki yeh class ka method hai
        #to scrap the data 
        #we will use the requests module to get the data from cricbuzz
        details = self.scrap()



        #need to parse and read the relevant data from the html page
        live_match = {} #dictionary to store the match details
        for detail in details: #for each match detail in the details
            summary = self.match_summary(detail)
            print(summary) #get the summary of the match
            if summary is not None :
                match_header = self.match_header(detail).text
                teams = self.teams_names(detail) #get the team names
        
                #print(teams) #print the team names
    
    #function to get the team names
    def teams_names(self, detail):
        t =[]
        team1_detail = detail.find("div", class_ = "cb-hmsch-bat-txt").text #get the team 1 name 
        print(team1_detail) 
     #function to get the match summary
    def match_summary(self, detail): #function to get the summary of the match
        return detail.find("div", class_= "cb-mtch-crd-state")
    
    #function to fetch the match header
    def match_header(self,detail): #function to get the header of the match
        return detail.find("div", class_ = "cb-mtch-crd-hdr")

    
    #function to scrap the website and get the data from cricbuzz
    def scrap(self):
        URL = "https://www.cricbuzz.com/" #url of the cricbuzz website
        page = requests.get(URL) #get the page from the url
        #print(page) #print the page to check if the page is fetched successfully

        #parse this response and fetch the relevant data
        soup = BeautifulSoup(page.content, "html.parser") #parse the page using BeautifulSoup
        results = soup.find(id = "match_menu_container") #find the match menu container in the page
        #print(result) #print the result to check if the data is fetched successfully
        scrap_results = results.find_all("li", class_ = "cb-match-card")
        return scrap_results
    



def main():
    #start the GUI  window
    rootWindow = Tk()


    #fetch the cricket score
    obj = cricketscore(rootWindow)
    #create the object of the class and pass the root window to it


    #keep on diplaying the gui the window to fetch the live updates
    rootWindow.mainloop()
    


#Execute the file
if __name__ == "__main__": #kuch function and kuch variable ko run karne ke liye
    #if the file is run directly then execute the main function
    main()
    #execute the main function to start the program
```
